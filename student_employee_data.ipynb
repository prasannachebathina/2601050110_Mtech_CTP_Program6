```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    course: str

class Employee:
    def __init__(self, name, job):
        self.name = name
        self.job = job

student = Student("Prasanna", "CSE")
employee = Employee("Rahul", "Developer")

print("Student:", student)
print("Employee:", employee.name, employee.job)
```
